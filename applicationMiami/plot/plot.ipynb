```python
import pandas as pd
import matplotlib.pyplot as plt

def load_df(path):
    try:
        df = pd.read_csv(path)
    except:
        df = pd.read_csv(path, header=None)
        df.columns = ["Day","median","ci_low","ci_high","truth"][:df.shape[1]]
    cols = df.columns.tolist()
    Day = "Day" if "Day" in cols else cols[0]
    lo  = "ci_low" if "ci_low" in cols else cols[2]
    hi  = "ci_high" if "ci_high" in cols else cols[3]
    med = "median" if "median" in cols else cols[1]
    tr  = "truth" if "truth" in cols else cols[-1]
    return df, Day, med, lo, hi, tr

# ---------- only Istar ----------
df, Day, med, lo, hi, tr = load_df("Istar_stats_with_truth_power.csv")

plt.figure(figsize=(6,4))
plt.fill_between(df[Day], df[lo], df[hi], alpha=0.25, label="95% CI")
plt.plot(df[Day], df[med], color="black", label="Median")
plt.plot(df[Day], df[tr], color="red", linestyle="--", label="Truth")
plt.xlabel("Day")
plt.ylabel(r"$I_t^*$")
plt.tight_layout()
plt.savefig("Istar_vs_truth_power.png", dpi=200)

```
